```python
import pandas as pd
dataset=pd.read_csv('./car_price_prediction.csv')
import numpy as np
dataset.head()

categorical_cols = ['Manufacturer', 'Model', 'Category', 'Leather interior', 'Fuel type', 'Gear box type', 'Drive wheels', 'Doors', 'Wheel', 'Color']
dataset = pd.get_dummies(dataset, columns=categorical_cols)

print(dataset.info())
print(dataset.describe())
print(dataset.isnull().sum())


# Remove any non-numeric characters (e.g., '2.0 Turbo' -> '2.0')
dataset['Engine volume'] = dataset['Engine volume'].str.extract('([0-9.]+)').astype(float)
dataset['Mileage'] = dataset['Mileage'].str.replace('[^0-9]', '', regex=True).astype(float)
dataset['Levy'] = dataset['Levy'].replace('-', np.nan).astype(float)

print(dataset.dtypes)
dataset.head()
```

```python
from sklearn.model_selection import train_test_split  # Correct import
from sklearn.preprocessing import StandardScaler

X=dataset.drop(columns=['Price'])
y=dataset['Price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

scaler= StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)

```

```python
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
# model training
model = HistGradientBoostingRegressor()
model.fit(X_train_scaled,y_train)
model_predictions=model.predict(X_test)

print(mean_absolute_error(y_test,model_predictions))
print(mean_squared_error(y_test,model_predictions))
print(r2_score(y_test,model_predictions))
```

```python
import joblib

joblib.dump(model, 'car-price-regressor.pkl')
joblib.dump(scaler, 'scaler.pkl')
```
